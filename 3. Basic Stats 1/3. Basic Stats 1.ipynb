Imort library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
pd.set_option("display.max_columns", 50)

df = pd.read_csv("sales_data_with_discounts.csv")
```

1. Load & Inspect

```python
print(df.shape)          # (450, 13)
print(df.info())
print(df.head())
print(df.isnull().sum())
print(df.duplicated().sum())
```

Descriptive Analytics for Numerical Columns

```python
num_cols = df.select_dtypes(include=np.number).columns.tolist()

summary = pd.DataFrame({
    "Mean":     df[num_cols].mean(),
    "Median":   df[num_cols].median(),
    "Mode":     df[num_cols].mode().iloc[0],
    "Std Dev":  df[num_cols].std(),
    "Skewness": df[num_cols].skew(),
    "Min":      df[num_cols].min(),
    "Max":      df[num_cols].max(),
    "IQR":      df[num_cols].quantile(.75) - df[num_cols].quantile(.25),
})
print(summary.round(2))
```

3. Data Visualization

3.1 Histograms

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
for ax, col in zip(axes.ravel(), num_cols):
    sns.histplot(df[col], bins=30, kde=True, ax=ax, color="steelblue")
    ax.set_title(f"Distribution of {col}")
    ax.axvline(df[col].mean(),   color="red",   ls="--", label="Mean")
    ax.axvline(df[col].median(), color="green", ls=":",  label="Median")
    ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

3.2 Box Plots

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
for ax, col in zip(axes.ravel(), num_cols):
    sns.boxplot(y=df[col], ax=ax, color="lightcoral")
    ax.set_title(f"Boxplot of {col}")
plt.tight_layout()
plt.show()
```

3.3 Bar Charts for Categorical Columns

```python
cat_cols = ["Day", "BU", "Brand", "SKU", "Model", "City"]

for col in cat_cols:
    plt.figure(figsize=(10, 4))
    order = df[col].value_counts().index
    sns.countplot(data=df, x=col, order=order, palette="viridis")
    plt.title(f"Frequency of each {col}")
    plt.xticks(rotation=90 if col in ("SKU", "Model") else 0)
    plt.tight_layout()
    plt.show()
```

4. Preprocessing

```python
# ---------- 4.1 Clean text & whitespace ----------
df.columns = df.columns.str.strip()
df["Model"] = df["Model"].str.strip()          # "YM-98 " , "Babaji Oil " had trailing spaces
df["Brand"] = df["Brand"].str.strip()
df["BU"]    = df["BU"].str.strip()

# ---------- 4.2 Parse dates ----------
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")
df["Month"]      = df["Date"].dt.month
df["DayOfWeek"]  = df["Date"].dt.dayofweek        # 0 = Monday
df["IsWeekend"]  = df["DayOfWeek"].isin([5, 6]).astype(int)

# ---------- 4.3 Drop useless / redundant columns ----------
df_clean = df.drop(columns=["City", "Model"])     # City: zero variance; Model: 1:1 with SKU

# ---------- 4.4 Integrity checks ----------
assert np.allclose(df["Total Sales Value"], df["Volume"] * df["Avg Price"], rtol=1e-6)
assert np.allclose(df["Net Sales Value"],
                   df["Total Sales Value"] - df["Discount Amount"], rtol=1e-6)

# ---------- 4.5 Outlier handling ----------
# Keep the premium-mobile outliers but tame the scale with a log transform
df_clean["Log_Total_Sales"] = np.log1p(df_clean["Total Sales Value"])
df_clean["Log_Avg_Price"]   = np.log1p(df_clean["Avg Price"])

# Alternative (if outliers must be capped rather than transformed):
# q1, q3 = df["Volume"].quantile([.25, .75]); iqr = q3 - q1
# df_clean["Volume_capped"] = df["Volume"].clip(q1 - 1.5*iqr, q3 + 1.5*iqr)

# ---------- 4.6 Encoding categoricals ----------
df_clean = pd.get_dummies(df_clean,
                          columns=["BU", "Brand"],
                          prefix=["BU", "Brand"],
                          drop_first=True)
# Day is ordinal -> map explicitly so it keeps its order
df_clean["Day_num"] = df_clean["Day"].map({
    "Monday": 1, "Tuesday": 2, "Wednesday": 3, "Thursday": 4,
    "Friday": 5, "Saturday": 6, "Sunday": 7})
df_clean = df_clean.drop(columns=["Day"])

# ---------- 4.7 Scaling ----------
from sklearn.preprocessing import StandardScaler

scale_cols = ["Volume", "Avg Price", "Discount Rate (%)"]
df_clean[[c + "_scaled" for c in scale_cols]] = \
    StandardScaler().fit_transform(df_clean[scale_cols])

print(df_clean.head())
print(df_clean.shape)      # 450 rows, engineered features added
```
