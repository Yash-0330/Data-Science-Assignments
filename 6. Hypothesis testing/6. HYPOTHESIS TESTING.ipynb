## HYPOTHESIS TESTING

Background:
Bombay hospitality Ltd. operates a franchise model for producing exotic Norwegian dinners throughout New England. The operating cost for a franchise in a week (W) is given by the equation W = $1,000 + $5X, where X represents the number of units produced in a week. Recent feedback from restaurant owners suggests that this cost model may no longer be accurate, as their observed weekly operating costs are higher.

Objective:
To investigate the restaurant owners' claim about the increase in weekly operating costs using hypothesis testing.

Data Provided:
●	The theoretical weekly operating cost model: W = $1,000 + $5X
●	Sample of 25 restaurants with a mean w
eekly cost of Rs. 3,050
●	Number of units produced in a week (X) follows a normal distribution with a mean (μ) of 600 units and a standard deviation (σ) of 25 units



1. State the Hypotheses statement:

```python
import numpy as np
from scipy import stats

# Given information
sample_mean = 3050
production_mean = 600
production_sd = 25
sample_size = 25
alpha = 0.05

# Step 1: Calculate theoretical mean cost
theoretical_mean = 1000 + 5 * production_mean

# Step 2: Calculate population SD of weekly cost
population_sd = 5 * production_sd

# Step 3: State hypotheses
print("H0: Population mean cost = 4000")
print("H1: Population mean cost > 4000")
```

2. Calculate the Test Statistic:

```python
import math

# Given values
sample_mean = 3050
X = 600
mu = 1000 + 5 * X
sigma = 5 * 25
n = 25

# Calculate Z-statistic
z = (sample_mean - mu) / (sigma / math.sqrt(n))

print("Sample Mean:", sample_mean)
print("Theoretical Mean:", mu)
print("Standard Deviation:", sigma)
print("Sample Size:", n)
print("Z-statistic:", z)
```

3. Determine the Probability and compare:

```python
from scipy.stats import norm

# Given values
z = -38.0
alpha = 0.05

# Right-tailed test
p_value = 1 - norm.cdf(z)

# Critical Z-value for 5% significance level
critical_z = norm.ppf(1 - alpha)

print("Z-statistic:", z)
print("P-value:", p_value)
print("Alpha level:", alpha)
print("Critical Z-value:", critical_z)

if p_value < alpha:
    print("Reject the null hypothesis (H0)")
else:
    print("Fail to reject the null hypothesis (H0)")
```

4. Make a Decision:

```python
# Step 6: Display results
import math
# Calculate standard error
standard_error = population_sd / math.sqrt(sample_size)

# Calculate Z-statistic
z_statistic = (sample_mean - theoretical_mean) / standard_error

# Critical Z-value
critical_value = 1.96
print("\n--- Hypothesis Testing Results ---")
print("Theoretical mean cost:", theoretical_mean)
print("Population standard deviation:", population_sd)
print("Standard error:", standard_error)
print("Z-statistic:", z_statistic)
print("Critical z-value:", round(critical_value, 4))

print("P-value:", p_value)

# Step 7: Make decision
if z_statistic > critical_value:
    print("\nDecision: Reject the null hypothesis.")
    print("There is evidence that weekly costs are higher.")
else:
    print("\nDecision: Fail to reject the null hypothesis.")
    print("No evidence supports the claim that weekly costs are higher.")
```

## ESTIMATION AND CONFIDENCE INTERVALS

Background

In quality control processes, especially when dealing with high-value items, destructive sampling is a necessary but costly method to ensure product quality. The test to determine whether an item meets the quality standards destroys the item, leading to the requirement of small sample sizes due to cost constraints.

Scenario

A manufacturer of print-heads for personal computers is interested in estimating the mean durability of their print-heads in terms of the number of characters printed before failure. To assess this, the manufacturer conducts a study on a small sample of print-heads due to the destructive nature of the testing process.

Data

A total of 15 print-heads were randomly selected and tested until failure. The durability of each print-head (in millions of characters) was recorded as follows:
1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29



a. Build 99% Confidence Interval Using Sample Standard Deviation

```python

import numpy as np
from scipy import stats

# Print-head durability data (in millions of characters)
data = np.array([
    1.13, 1.55, 1.43, 0.92, 1.25,
    1.36, 1.32, 0.85, 1.07, 1.48,
    1.20, 1.33, 1.18, 1.22, 1.29
])

# Basic calculations
n = len(data)
mean = np.mean(data)
sample_sd = np.std(data, ddof=1)
confidence = 0.99
alpha = 1 - confidence

print("Sample size:", n)
print("Sample mean:", round(mean, 4))
print("Sample standard deviation:", round(sample_sd, 4))

# Part (a): 99% Confidence Interval using t-distribution
t_critical = stats.t.ppf(1 - alpha / 2, df=n - 1)
margin_t = t_critical * sample_sd / np.sqrt(n)

lower_t = mean - margin_t
upper_t = mean + margin_t

print("\nPart (a): t-distribution")
print("Critical t-value:", round(t_critical, 4))
print("Margin of error:", round(margin_t, 4))
print("99% Confidence Interval:",
      (round(lower_t, 4), round(upper_t, 4)))
```

b. Build 99% Confidence Interval Using Known Population Standard Deviation

```python

import numpy as np
from scipy import stats

data = np.array([
    1.13, 1.55, 1.43, 0.92, 1.25,
    1.36, 1.32, 0.85, 1.07, 1.48,
    1.20, 1.33, 1.18, 1.22, 1.29
])

population_sd = 0.2
z_critical = stats.norm.ppf(1 - alpha / 2)
margin_z = z_critical * population_sd / np.sqrt(n)

lower_z = mean - margin_z
upper_z = mean + margin_z

print("\nPart (b): z-distribution")
print("Critical z-value:", round(z_critical, 4))
print("Margin of error:", round(margin_z, 4))
print("99% Confidence Interval:",
      (round(lower_z, 4), round(upper_z, 4)))
```
